```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

tokens = encoding.encode(
    "Hi, Do you know the health benefits of coffee if consume regularly?"
)
print(tokens)
```

```python
for token_id in tokens:
    text = encoding.decode(tokens=[token_id])
    print(f"{token_id} = '{text}'")
```

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)

BASE_URL = os.getenv("OPENROUTER_API_URL")
API_KEY = os.getenv("OPENROUTER_API_KEY")
```

```python
from openai import OpenAI

openai = OpenAI(base_url=BASE_URL, api_key=API_KEY)
```

```python
messages = [
    {
        "role": "user",
        "content": "Hey, I am just completed my B.Tech in computer science.",
    },
]

response = openai.chat.completions.create(
    messages=messages,  # type: ignore
    model="google/gemma-3-4b-it:free",
)
messages.append({"role": "assistant", "content": response.choices[0].message.content})  # type: ignore
print(response.choices[0].message.content)
```

```python
messages.append(
    {
        "role": "user",
        "content": "Where should i apply for job for my degree?",
    }
)

response = openai.chat.completions.create(
    messages=messages,  # type: ignore
    model="google/gemma-3-4b-it:free",
)
print(response.choices[0].message.content)
```
