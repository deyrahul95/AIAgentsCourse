### Basic Open AI Model config 

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)

LLAMA_BASE_URL = os.getenv("LLAMA_BASE_URL")
LLAMA_MODEL = "openai/qwen"

if LLAMA_BASE_URL is not None:
    print(f"LLAMA BASE URL -> {LLAMA_BASE_URL}")
    print(f"LLAMA MODEL -> {LLAMA_MODEL}")
```

```python
from openai import OpenAI
from openai.types.chat.chat_completion_message_param import ChatCompletionMessageParam

llama_client = OpenAI(base_url=LLAMA_BASE_URL, api_key="NotNeeded")
```

```python
def chat_llama(query: str) -> str | None:
    messages: list[ChatCompletionMessageParam] = [
        {"role": "system", "content": "You are a helpful assistant"},
        {"role": "user", "content": query},
    ]
    response = llama_client.chat.completions.create(
        model=LLAMA_MODEL, messages=messages
    )
    return response.choices[0].message.content
```

```python
chat_llama("What is today's date")
```

### User Interface

```python
import gradio as gr
```

```python
gr.Interface(
    fn=chat_llama,
    inputs="textbox",
    outputs="textbox",
    flagging_mode="never",
).launch()
```
