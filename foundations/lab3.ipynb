```python
from dotenv import load_dotenv
from openai import OpenAI
import os

load_dotenv(override=True)
```

```python
openai = OpenAI(
    base_url=os.getenv("OPENROUTER_API_URL"), api_key=os.getenv("OPENROUTER_API_KEY")
)
model = "openai/gpt-oss-120b:free"
```

```python
from pydantic import BaseModel, Field
# from typing import Optional


class LLMOutput(BaseModel):
    """This model represent the output structure for the LLM call"""

    query: str = Field(..., description="The question query stored here.")
    answer: str = Field(..., description="LLM response on this user query stored here.")
    # code: Optional[str] = Field(default=None, description="LLM code output store here.")
```

```python
# system_prompt = f"Your are a helpful coding assistance to help user resolved their queries. If the user asked for coding question please give the code and explanation in clear and easy to understand language. Always use the response format {LLMOutput} to give response"
query = "What is the capital of India?"

response = openai.beta.chat.completions.parse(
    model=model,
    messages=[
        # {"role": "system", "content": system_prompt},
        {"role": "user", "content": query},
    ],
    response_format=LLMOutput,
)
print(response.choices[0].message.parsed)
```
